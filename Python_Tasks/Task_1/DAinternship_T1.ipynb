```python
#35
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
#[[1,99],[3,4]] --here shallow copy is created
# and it shares inner list
```

```python
#36
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
#[[1, 2], [3, 4]] # deepcopy creates a completly diff copy of existing element
```
