```python
#Q10
# take a string as input
# create an empty dictionary to store character frequencies
# use a for loop to check each character
# ignore spaces
# if the character exists in the dictionary, increase its count by 1, otherwise set it to 1
# create an empty list to store repeated characters
# loop through the dictionary to check the frequency of each character
# if the frequency is greater than 1, append the character to the list
# print the list of repeated characters
word="programming"
count={}
for i in word:
    if i !=" ":
        if i in count:
            count[i] += 1
        else:
            count[i] = 1

char=[]

for i in count:
    if count[i]>1:
        char.append(i)

print(char)
```

```python
#Q16
# convert the sentence to lowercase using lower() and split it in words
# create an empty dictionary 
# use a for loop to check each word
# if the word exists in the dictionary, increase its count by 1, otherwise set it to 1
# initialize max_count = 0 and most_frequent = ""
# loop through the dictionary and compare word frequencies
# if a higher frequency is found, update both variables
# if frequencies are equal, keep the first encountered word
# print the most frequent word

sent = "Python is easy and Python is powerful"
words = sent.lower().split()
freq = {}
for i in words:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

max_count = 0
most_frequent = ""
for i in freq:
    if freq[i] > max_count:
        max_count = freq[i]
        most_frequent = i

print(most_frequent)

```

```python
#Q17
# convert the sentence to lowercase using lower() and split the words
# split the sentence into individual words
# create an empty dictionary to store word frequencies
# use a for loop to check each word
# if the word exists in the dictionary, increase its count by 1, otherwise set it to 1
# create an empty list to store repeated words
# loop through the dictionary to check the frequency of each word
# if the frequency is greater than 1, append the word to the list
# print the list of repeated words
sent = "data science uses data and science tools"
words = sent.lower().split()
freq = {}
for i in words:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1
repeated = []
for i in freq:
    if freq[i] > 1:
        repeated.append(i)

print(repeated)
```

```python
#Q18

# convert the sentence to lowercase using lower() and split it
# create an empty dictionary
# use a for loop to check each word
# if the word exists in the dictionary, increase its count by 1, otherwise set it to 1
# initialize first_unique = None
# loop through the dictionary to check the frequency of each word
# if the frequency is equal to 1, store that word in first_unique
# use break to stop the loop after finding the first unique word
# print the first unique word

sent = "apple banana apple mango banana orange"
words = sent.lower().split()
freq = {}
for i in words:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1
first_unique = None
for i in freq:
    if freq[i] == 1:
        first_unique = i
        break

print(first_unique)
```

```python
#Q20
# convert the sentence to lowercase using lower()and split it
# create an empty dictionary to store word frequencies
# use a for loop to check each word
# if the word exists in the dictionary, increase its count by 1, otherwise set it to 1
# initialize longest = "" to store the longest unique word
# loop through the dictionary to check each word
# if the word appears exactly once, compare its length with longest
# if the current word is longer, update longest
# if lengths are equal, keep the first encountered word
# print the longest unique word
sent = "Python makes coding useful Python"
words = sent.lower().split()
freq = {}
for i in words:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1
longest = ""
for i in freq:
    if freq[i] == 1:
        if len(i) > len(longest):
            longest = i

print(longest)
```
