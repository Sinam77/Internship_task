Q1

1. Take a sentence as input.
2. Split the sentence into words.
3. Create an empty dictionary.
4. Go through each word.
5. If the word exists, increase its count.
6. Otherwise, add the word with count 1.
7. Return the dictionary.

Q2

1. Take a word as input.
2. Create an empty dictionary.
3. Go through each letter.
4. If the letter exists, increase its count.
5. Otherwise, add the letter with count 1.
6. Return the dictionary.

Q3

1. Take a word as input
2. Create an empty dictionary
3. Count the frequency of each character
4. Traverse the word again
5. Check whether the frequency of a character is 1
6. If yes, return that character immediately
7. If no such character exists, return None

Q4

1. Take two strings
2. Create two empty dictionaries.
3. Count the characters in the first and second string.
4. Store their frequencies in the first and second dictionary.
5. Compare both dictionaries
6. Return True if they match, otherwise False.

Q5

1. Take a string as input.
2. Create an empty set or list.
3. Go through each character in the string.
4. Check if the character is already in the set or list.
5. If yes, return that character.
6. Otherwise, add it to the set.
7. If no character repeats, return None.

Q6

1. Take a list of numbers.
2. Create two empty lists.
3. Go through each number in the original list.
4. If the number is zero, add it to the zeros list.
5. Otherwise, add it to the non-zero list.
6. Combine the non-zero list with the zeros list.
7. Return the combined list.

Q7

1. Take a list of numbers.
2. Remove duplicate numbers using a set.
3. Check if at least two distinct numbers exist.
4. If not, return None.
5. Sort the unique numbers from highest to lowest.
6. Return the number at index 1.

Q8

1. Take a list as input.
2. Create an empty list.
3. Iterate through the original list.
4. Check if the element is absent from the new list.
5. If absent, append it.
6. Return the new list.

Q9

1. Take a list and N as input.
2. Calculate the expected sum from 1 to N.
3. Calculate the sum of the given list.
4. Subtract the actual sum from the expected sum.
5. Return the missing number.

Q10
1. Take a list and target as input.
2. Start checking the second element from left to right.
3. For each second element, check all earlier elements.
4. Calculate the sum of each pair.
5. If the sum equals the target, return the pair.
6. If no pair is found, return an empty list.

Q11

1. Take a non-empty list.
2. Assume the first number is closest to zero.
3. Compare the absolute value of each number.
4. Update the closest number if a smaller distance is found.
5. If distances are equal, choose the larger number.
6. Return the closest number.

Q12

1. Take two lists.
2. Create an empty result list.
3. Iterate through the first list.
4. Check if each element exists in the second list.
5. Also check that it is not already in the result.
6. If both conditions are true, append it.
7. Return the result.

Q13

1. Take a list and K as input.
2. If the list is empty, return an empty list.
3. Calculate K modulo the list length.
4. If K is zero, return a copy of the list.
5. Take the last K elements.
6. Take the remaining elements from the beginning.
7. Combine both parts and return the result.

Q14

1. Take a list containing at least two numbers.
2. Assume the sum of the first two elements is maximum.
3. Iterate through consecutive pairs.
4. Calculate each pair's sum.
5. If the current sum is greater, update the maximum.
6. Return the maximum sum.

Q15

1. Take a list and target as input.
2. Create an empty result list.
3. Use nested loops to check every pair of different positions.
4. Check whether the pair's sum equals the target.
5. If yes, sort the two numbers in ascending order.
6. Check whether the pair is already in the result.
7. If not, append it.
8. Sort the complete result.
9. Return the list of pairs.
